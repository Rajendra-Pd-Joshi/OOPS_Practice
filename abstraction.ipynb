# Abstraction in Python

**Abstraction** is the process of **hiding implementation details and exposing only the essential functionality** to the user.

In simple words:

> **Abstraction tells the user what an object can do without showing how it does it.**

It makes programs easier to understand and interact with because users only need to know the necessary functionality.

---

# Real-Life Example

Consider a **graphics application** containing different shapes:

* Circle
* Rectangle
* Triangle

All shapes may have some common data:

* `name`
* `color`
* `border_thickness`
* `fill_style`

They may also support common operations:

```python id="ib6f7w"
draw()
resize()
get_area()
```

However, each shape performs these operations differently.

For example:

* **Circle** calculates area using its radius.
* **Rectangle** calculates area using length and width.
* **Triangle** calculates area using base and height.

The user only needs to know:

```python id="ed8zha"
shape.draw()
shape.get_area()
```

The internal calculation is hidden.

This is **abstraction**.

# Abstract Base Class (ABC)

Python provides the `abc` module for implementing abstraction.

`ABC` stands for:

> **Abstract Base Class**

An abstract class acts as a **blueprint for other classes**.

It can define methods that subclasses are required to implement.

Import it using:

```python id="ilj6ge"
from abc import ABC, abstractmethod
```

---

# Basic Example

```python id="d67lxy"
from abc import ABC, abstractmethod


class Greet(ABC):

    @abstractmethod
    def say_hello(self):
        pass


class English(Greet):

    def say_hello(self):
        return "Hello!"


g = English()

print(g.say_hello())
```

### Output

```text id="eml12f"
Hello!
```

### Explanation

`Greet` is an abstract class:

```python id="snizqg"
class Greet(ABC):
```

It contains an abstract method:

```python id="u44pyq"
@abstractmethod
def say_hello(self):
    pass
```

This tells subclasses:

> **You must provide an implementation of `say_hello()`.**

The `English` class provides that implementation:

```python id="exmh7j"
def say_hello(self):
    return "Hello!"
```

Therefore, `English` becomes a concrete class and can be instantiated.

# Components of Abstraction

The major components are:

1. Abstract Class
2. Abstract Method
3. Concrete Method
4. Abstract Property
5. Concrete Class

---

# 1. Abstract Class

An **abstract class** is a class designed primarily to act as a blueprint for other classes.

It usually contains one or more abstract methods.

### Example

```python id="evsc48"
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass
```

Here:

```python id="8kud6c"
Animal
```

is an abstract class.

A class becomes abstract when it inherits from `ABC` and has abstract methods or properties that remain unimplemented.

---

# 2. Abstract Method

An **abstract method** is a method that subclasses are required to implement.

It is created using:

```python id="48i7rz"
@abstractmethod
```

### Example

```python id="twgj8h"
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass
```

Here:

```python id="o3l9n5"
make_sound()
```

is an abstract method.

The parent class defines **what operation must exist**, while the subclass decides **how that operation works**.

For example:

```python id="1g6bsj"
class Dog(Animal):

    def make_sound(self):
        return "Bark"
```

Now `Dog` has implemented the required method.

---

# 3. Concrete Method

A **concrete method** is a normal method with a complete implementation.

An abstract class can contain both:

* Abstract methods
* Concrete methods

Subclasses can directly use inherited concrete methods.

### Example

```python id="v6k5gy"
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

    def move(self):
        return "Moving"


class Dog(Animal):

    def make_sound(self):
        return "Bark"


dog = Dog()

print(dog.move())
```

### Output

```text id="3k9oaz"
Moving
```

### Explanation

The method:

```python id="j23vcq"
def move(self):
    return "Moving"
```

is a **concrete method**.

It already contains an implementation.

Therefore, `Dog` can inherit and use it directly:

```python id="7tmwva"
dog.move()
```

without overriding it.

---

# Abstract Method vs Concrete Method

| Abstract Method                               | Concrete Method                   |
| --------------------------------------------- | --------------------------------- |
| Marked with `@abstractmethod`                 | Normal method                     |
| Subclass must implement it to become concrete | Subclass can use it directly      |
| Defines required behavior/interface           | Provides actual reusable behavior |
| Often has no useful implementation            | Has a complete implementation     |

---

# 4. Abstract Properties

Python can also define **abstract properties**.

An abstract property requires subclasses to provide an implementation for a property.

It commonly uses:

```python id="gmh3s1"
@property
@abstractmethod
```

### Example

```python id="6xipom"
from abc import ABC, abstractmethod


class Animal(ABC):

    @property
    @abstractmethod
    def species(self):
        pass


class Dog(Animal):

    @property
    def species(self):
        return "Canine"


dog = Dog()

print(dog.species)
```

### Output

```text id="9ot3lv"
Canine
```

### Explanation

The parent class defines:

```python id="dqwlgi"
@property
@abstractmethod
def species(self):
    pass
```

Therefore, subclasses must provide an implementation of the `species` property.

`Dog` implements it:

```python id="qczymv"
@property
def species(self):
    return "Canine"
```

Now:

```python id="9pumxn"
dog.species
```

returns:

```text id="a2y8l6"
Canine
```

---

# 5. Abstract Class Instantiation

An abstract class with unimplemented abstract methods **cannot be instantiated directly**.

### Example

```python id="nmqqh4"
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


animal = Animal()
```

### Output

Python raises a `TypeError`, similar to:

```text id="b2vp7k"
TypeError: Can't instantiate abstract class Animal
with abstract method make_sound
```

### Why?

`Animal` defines:

```python id="38f0d1"
@abstractmethod
def make_sound(self):
    pass
```

but does not provide the required concrete implementation.

Therefore:

```python id="bsmvkt"
Animal()
```

cannot be created directly.

---

# Concrete Subclass

A subclass that implements all required abstract methods becomes a **concrete class**.

### Example

```python id="4i7tmw"
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


class Dog(Animal):

    def make_sound(self):
        return "Bark"


dog = Dog()

print(dog.make_sound())
```

### Output

```text id="4s8oei"
Bark
```

The relationship is:

```text id="zn7b5k"
Animal
(Abstract Class)
     │
     │ requires make_sound()
     ↓
Dog
(Concrete Class)
     │
     │ implements make_sound()
     ↓
Object can be created
```

# Complete Example of Abstraction

```python id="6kgnvp"
from abc import ABC, abstractmethod


class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    def display(self):
        print("This is a shape")


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


circle = Circle(5)
rectangle = Rectangle(4, 6)

print(circle.area())
print(rectangle.area())

circle.display()
```

### Output

```text id="gtq3m2"
78.5
24
This is a shape
```

### Explanation

`Shape` defines the common interface:

```python id="3b83h6"
@abstractmethod
def area(self):
    pass
```

Every concrete shape must implement `area()`.

`Circle` implements:

```python id="gnj5v9"
def area(self):
    return 3.14 * self.radius ** 2
```

`Rectangle` implements:

```python id="f7x7dz"
def area(self):
    return self.length * self.width
```

The user simply calls:

```python id="9q4suv"
circle.area()
rectangle.area()
```

without needing to worry about the internal calculation.

This is the main idea of **abstraction**.

# Why Do We Need Abstraction?

### 1. Hides Complexity

Users don't need to understand internal implementation details.

### 2. Provides a Clear Interface

Users only interact with the functionality they need.

### 3. Improves Maintainability

Internal implementation can change without necessarily changing how outside code interacts with the class.

### 4. Enforces Structure

Abstract methods ensure that subclasses implement required behavior.

### 5. Promotes Code Reusability

Concrete methods in abstract classes can be shared among subclasses.

### 6. Supports Polymorphism

Different subclasses can provide different implementations of the same abstract method.

---

# Abstraction vs Encapsulation

These concepts are related but should not be confused.

| Abstraction                                   | Encapsulation                                                                      |
| --------------------------------------------- | ---------------------------------------------------------------------------------- |
| Hides implementation complexity               | Controls access to data and implementation                                         |
| Focuses on **what an object does**            | Focuses on bundling and protecting internal state                                  |
| Exposes essential functionality               | Restricts/directs access to members                                                |
| Commonly implemented with ABCs and interfaces | Commonly implemented with classes, naming conventions, properties, getters/setters |

### Easy Way to Remember

> **Abstraction = What should the user know?**

> **Encapsulation = How should internal data be accessed and protected?**
