# Encapsulation in Python

**Encapsulation** is one of the core concepts of **Object-Oriented Programming (OOP)**.

Encapsulation means **binding data members and methods together into a single unit**, usually a class, while controlling how the internal data is accessed and modified.

## Key Benefits of Encapsulation

* Helps improve **maintainability, readability, and usability**.
* Helps maintain **data integrity** by controlling how values are assigned.
* Allows **validation** before modifying data.
* Hides internal implementation details.
* Helps achieve **abstraction**.
* Allows internal implementation to change without affecting code that uses the class.

---

# Basic Example of Encapsulation

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name          # Public attribute
        self.__salary = salary    # Private attribute


emp = Employee("Fedrick", 50000)

print(emp.name)
print(emp.__salary)
```

### Output

```text
Fedrick

AttributeError: 'Employee' object has no attribute '__salary'
```

### Explanation

* `self.name` is a **public attribute**, so it can be accessed directly.
* `self.__salary` is a **private attribute**.
* `emp.name` works because `name` is public.
* `emp.__salary` raises an `AttributeError` because `__salary` is private.

---

# Why Do We Need Encapsulation?

Encapsulation is useful because it:

### 1. Protects Data

It helps prevent accidental modification or inappropriate access to internal data.

### 2. Provides Controlled Access

Getter and setter methods can control how data is read or modified.

### 3. Allows Validation

We can validate data before storing it.

For example:

```python
if salary > 0:
    self.__salary = salary
```

### 4. Improves Modularity

Internal implementation details can be hidden inside the class.

### 5. Simplifies Maintenance

Data-handling logic can be centralized inside the class.

### 6. Represents Real-World Scenarios

For example, a bank account may not allow users to directly modify their balance.

Instead, operations such as:

```python
account.deposit(500)
account.withdraw(200)
```

can control how the balance changes.

---

# Access Specifiers in Python

Access specifiers describe how class members are intended to be accessed.

Python commonly uses three levels:

1. **Public**
2. **Protected**
3. **Private**

> **Important:** Python does not enforce `public`, `protected`, and `private` access modifiers as strictly as languages such as Java or C++. Python mainly uses naming conventions and **name mangling**.

---

# 1. Public Members

Public members can be accessed from:

* Inside the class
* Outside the class
* Other modules

By default, members in Python are public.

They are defined **without an underscore prefix**.

### Example

```python
class Employee:
    def __init__(self, name):
        self.name = name       # Public attribute

    def display_name(self):    # Public method
        print(self.name)


emp = Employee("John")

emp.display_name()
print(emp.name)
```

### Output

```text
John
John
```

### Explanation

```python
self.name
```

is public because it does not start with `_` or `__`.

Similarly:

```python
display_name()
```

is a public method.

Both can be accessed directly using the object.

---

# 2. Protected Members

Protected members are intended to be used:

* Inside the class
* Inside subclasses
* Within the class hierarchy

They are defined using a **single underscore (`_`)**.

Example:

```python
self._age
```

### Example

```python
class Employee:
    def __init__(self, name, age):
        self.name = name       # Public
        self._age = age        # Protected


class SubEmployee(Employee):
    def show_age(self):
        print("Age:", self._age)


emp = SubEmployee("Ross", 30)

print(emp.name)
emp.show_age()
```

### Output

```text
Ross
Age: 30
```

### Explanation

* `self.name` is public.
* `self._age` is intended to be protected.
* `SubEmployee` can access `_age` because it inherits from `Employee`.
* Python does **not strictly prevent** outside code from accessing `_age`.

For example, this technically works:

```python
print(emp._age)
```

But the underscore tells other programmers:

> **"This member is intended for internal use."**

---

# 3. Private Members

Private members are intended to be accessed only from within the class.

They are defined using a **double underscore (`__`)**.

Example:

```python
self.__salary
```

Python uses a mechanism called **name mangling** for private members.

For example:

```python
__salary
```

inside:

```python
Employee
```

is internally changed approximately to:

```python
_Employee__salary
```

This discourages direct access from outside the class.

> Python's private members are not truly private in the same way as some other languages. Name mangling mainly prevents accidental access and naming conflicts.

---

## Example of a Private Attribute

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def show_salary(self):
        print("Salary:", self.__salary)


emp = Employee("Robert", 60000)

print(emp.name)
emp.show_salary()

# print(emp.__salary)  # AttributeError
```

### Output

```text
Robert
Salary: 60000
```

### Explanation

* `name` is public.
* `__salary` is private.
* `show_salary()` is public and provides controlled access to `__salary`.
* Directly using `emp.__salary` results in an `AttributeError`.

---

# Public vs Protected vs Private

| Type      | Syntax   | Intended Access      |
| --------- | -------- | -------------------- |
| Public    | `name`   | Anywhere             |
| Protected | `_name`  | Class and subclasses |
| Private   | `__name` | Inside the class     |

### Easy Way to Remember

```text
name      → Public
_name     → Protected
__name    → Private
```

---

# Declaring Protected and Private Methods

The same naming conventions can be applied to methods.

### Protected Method

Use a single underscore:

```python
def _show_balance(self):
    ...
```

### Private Method

Use a double underscore:

```python
def __update_balance(self):
    ...
```

---

# Example: Protected and Private Methods

```python
class BankAccount:
    def __init__(self):
        self.balance = 1000

    def _show_balance(self):
        print(f"Balance: ₹{self.balance}")

    def __update_balance(self, amount):
        self.balance += amount

    def deposit(self, amount):
        if amount > 0:
            self.__update_balance(amount)
            self._show_balance()
        else:
            print("Invalid deposit amount!")


account = BankAccount()

account._show_balance()

# account.__update_balance(500)  # AttributeError

account.deposit(500)
```

### Output

```text
Balance: ₹1000
Balance: ₹1500
```

### Explanation

#### `_show_balance()`

This is a **protected method**.

It can technically be accessed from outside:

```python
account._show_balance()
```

but it is intended for internal or subclass use.

#### `__update_balance()`

This is a **private method**.

It is intended to be accessed from inside the class.

#### `deposit()`

This is a public method that provides controlled access to the internal functionality.

The flow is:

```text
deposit()
   ↓
__update_balance()
   ↓
_show_balance()
```

This is an example of encapsulation because the user interacts with the public method while internal operations remain hidden.

# Getter and Setter Methods

**Getter** and **setter** methods are commonly used to provide controlled access to private attributes.

### Getter

A getter is used to **read** a value.

```python
def get_salary(self):
    return self.__salary
```

### Setter

A setter is used to **modify** a value.

```python
def set_salary(self, amount):
    if amount > 0:
        self.__salary = amount
```

Setters are particularly useful when we need to perform validation.

---

# Example: Getter and Setter

```python
class Employee:
    def __init__(self):
        self.__salary = 50000

    def get_salary(self):
        return self.__salary

    def set_salary(self, amount):
        if amount > 0:
            self.__salary = amount
        else:
            print("Invalid salary amount!")


emp = Employee()

print(emp.get_salary())

emp.set_salary(60000)

print(emp.get_salary())
```

### Output

```text
50000
60000
```

### Explanation

`__salary` is private:

```python
self.__salary = 50000
```

The outside code does not directly modify it.

Instead, it uses:

```python
emp.get_salary()
```

to read the value and:

```python
emp.set_salary(60000)
```

to update it.

The setter can also validate the value:

```python
if amount > 0:
```

Therefore, invalid values can be rejected.

---

# Encapsulation Flow

A simple way to understand encapsulation is:

```text
        Outside Code
             │
             ▼
     Public Method
             │
             ▼
    Validation / Logic
             │
             ▼
    Private Attribute
```

For example:

```text
emp.set_salary(60000)
        ↓
   Validate amount
        ↓
   __salary = 60000
```

The outside code does not need to know how the salary is stored internally.

# Encapsulation and Abstraction

Encapsulation and abstraction are related but different concepts.

| Encapsulation                                      | Abstraction                                   |
| -------------------------------------------------- | --------------------------------------------- |
| Bundles data and methods together                  | Hides unnecessary implementation details      |
| Controls access to data                            | Shows only essential functionality            |
| Focuses on **data protection/access**              | Focuses on **what to expose**                 |
| Uses classes, naming conventions, properties, etc. | Often uses abstract classes, interfaces, etc. |

### Simple Example

In a bank account:

```python
account.deposit(500)
```

The user knows **what** operation is available.

They do not need to know **how** the balance is internally updated.

That combines the ideas of encapsulation and abstraction.

---

# Important Python Note

Python does **not** have strict access modifiers like:

```text
public
private
protected
```

Instead, Python uses conventions:

```python
name       # Public
_name      # Protected convention
__name     # Private through name mangling
```

Therefore:

> **Python relies on conventions and name mangling rather than strict access restrictions.**
